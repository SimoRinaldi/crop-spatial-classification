```python
from pathlib import Path

main_directory = Path("crops_types_yearly_capitanata")

tifs = {}

for d in main_directory.iterdir():
    if d.is_dir():
        year = d.name
        
        lista_tifs = list(d.rglob("*.tif"))
        
        tifs[year] = [str(tif) for tif in lista_tifs]

print("Anni trovati:", list(tifs.keys()))

tif_2017 = tifs.get("2017", [])
print(f"\nTrovati {len(tif_2017)} file .tif per il 2017:")
for file_tif in tif_2017:
    print(file_tif)
```

```python
import rioxarray

# File GeoTIFF originale (EPSG:3035)
raster = rioxarray.open_rasterio("file_originale_3035.tif")

# Riproiezione dell'intero raster in EPSG:4326
raster_4326 = raster.rio.reproject("EPSG:4326")

# Salvataggio del risultato in un nuovo file .tif
raster_4326.rio.to_raster("file_convertito_4326.tif")
```
